```r
library(castor)
library(ape)
library(phytools)
```

```r
tree_bac <- read.tree('../data/tree/bac120_r207_selected.tree')
```

```r
dat<-read.csv("../data/csv/OGT.csv",header=T)
```

```r
dat2 <- dat[dat["domain"]=="Bacteria",]
dat3 <- dat2[,c("accession","OGT")]
colnames(dat3) <- c("label","OGT")
rownames(dat3) <- dat3[,1]
dat_bac <- dat3[tree_bac$tip,]
tip_states=dat_bac[,"OGT"]
```

```r
result=asr_squared_change_parsimony(tree_bac, tip_states ,weighted=TRUE, check_input=TRUE)
```

```r
dat_genus <- dat2[,c("accession","genus","OGT")]
```

```r
l=unique(dat_genus[,"genus"])
```

```r
createEmptyDf = function( nrow, ncol, colnames = c() ){
  if( missing( ncol ) && length( colnames ) > 0 ){
    ncol = length( colnames )
  }
  data.frame( matrix( vector(), nrow, ncol, dimnames = list( c(), colnames ) ) )
}
bac_data = createEmptyDf(length(l), colnames = c( "genus", "OGT" ) )
```

```r
i=1
for (genus in l){
    temp=dat_genus[dat_genus["genus"]==genus,]
    index=findMRCA(tree_bac, temp[,"accession"])
    if(is.null(index)){
        next 
    }
    OGT=result$ancestral_states[index-2869]
    bac_data[i,]=c(genus,OGT)
    i=i+1
}
```

```r
bac_data=bac_data[1:i-1,]
```
